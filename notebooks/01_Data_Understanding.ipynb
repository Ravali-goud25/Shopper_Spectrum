## Quantity Validation

- After removing cancelled invoices, no records with Quantity ≤ 0 were found.
- Therefore, no additional quantity cleaning was required.

```python
df[df["UnitPrice"] <= 0].shape
```

```python
df[df["UnitPrice"] <= 0].head()
```

```python
df = df[df["UnitPrice"] > 0]
```

```python
df[df["UnitPrice"] <= 0].shape
```

## Unit Price Validation

- 40 records contained UnitPrice ≤ 0.
- These records do not represent valid sales transactions.
- Such records can affect revenue calculations and customer monetary analysis.
- Therefore, all records with UnitPrice ≤ 0 were removed.

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

## Revenue Feature Engineering

A new column named Revenue was created using:

Revenue = Quantity × UnitPrice

This column represents the total sales amount generated by each transaction and will be used for:

- Revenue Analysis
- Customer Monetary Value Calculation
- RFM Analysis
- Customer Segmentation
- Business Insights

# Data Cleaning Summary

Initial Dataset Size:
- Rows: 541,909
- Columns: 8

Data Cleaning Steps Performed:
1. Removed rows with missing CustomerID.
2. Removed duplicate records.
3. Removed cancelled invoices (InvoiceNo starting with 'C').
4. Verified no invalid quantities remained.
5. Removed records with UnitPrice ≤ 0.
6. Converted InvoiceDate to datetime format.
7. Created Revenue column using Quantity × UnitPrice.

Final Dataset Size:
- Rows: 392,692
- Columns: 9

The cleaned dataset is ready for Exploratory Data Analysis (EDA), RFM Analysis, Customer Segmentation, and Product Recommendation.
